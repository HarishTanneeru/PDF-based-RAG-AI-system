```python
from langchain_openai import ChatOpenAI
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import OpenAIEmbeddings
```

## STEP-1 : Load the PDF into Text

```python
text_data= PyPDFLoader("NovaS.pdf").load()
for page in text_data:
    page.metadata["author"]="Harish"
text_data
```

## STEP-2 Splitting the data


```python
splitter= RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)
chunks= splitter.split_documents(text_data)
chunks
```

## Step-3 Creating the Embeddings and storing in vector DB

```python
embed_model= OpenAIEmbeddings(model="text-embedding-3-small")
embedded_chunks= embed_model.embed_documents([i.page_content for i in chunks])
len(embedded_chunks)
```
